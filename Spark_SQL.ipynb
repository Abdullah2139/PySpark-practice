```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("SparkSQL").master("local[*]").getOrCreate()
```

```python
spark.sql("SELECT 1 + 1").show()
```

```python
spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .csv("data/2010-12-01.csv")\
    .createOrReplaceTempView("some_sql_view")
```

```python
spark.sql("DESCRIBE some_sql_view").show()
```

```python
spark.sql("""
    SELECT InvoiceNo, SUM(UnitPrice) AS total_unit_price
    FROM some_sql_view
    GROUP BY InvoiceNo
""").show(5)
```

## Creating Managed Tables

```python
spark.sql("DROP TABLE IF EXISTS rn_data")

spark.sql("""
CREATE TABLE rn_data (
    InvoiceNo STRING,
    StockCode STRING,
    Description STRING,
    Quantity INT,
    InvoiceDate STRING,
    UnitPrice DOUBLE,
    CustomerID STRING,
    Country STRING
)
USING CSV
OPTIONS (
    path 'C:/Users/97156/Downloads/pysparkPractice/data/2010-12-01.csv',
    header 'true'
)
""")
```

```python
spark.sql("SELECT * FROM rn_data LIMIT 10").show()
```

## Creating External Tables

```python
spark.sql("""
CREATE EXTERNAL TABLE IF NOT EXISTS rn_data_1(
    InvoiceNo STRING,
    StockCode STRING,
    Description STRING,
    Quantity INT,
    InvoiceDate STRING,
    UnitPrice DOUBLE,
    CustomerID STRING,
    Country STRING
) USING CSV 
OPTIONS (
    path 'C:/Users/97156/Downloads/pysparkPractice/data/2010-12-01.csv',
    header 'true'
    )
""")
```

```python
spark.sql("SELECT * FROM rn_data_1").show(2)
```
